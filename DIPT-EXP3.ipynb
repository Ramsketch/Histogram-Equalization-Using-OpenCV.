```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read Grayscale Image
img = cv2.imread('parrot.webp', cv2.IMREAD_GRAYSCALE)

# Display Original Grayscale Image
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.show()

# Original Histogram
plt.hist(img.ravel(), 256, range=[0, 256])
plt.title('Original Histogram')
plt.show()

# Histogram Equalization
img_eq = cv2.equalizeHist(img)

# Equalized Histogram
plt.hist(img_eq.ravel(), 256, range=[0, 256])
plt.title('Equalized Histogram')
plt.show()

# Display Equalized Image
plt.imshow(img_eq, cmap='gray')
plt.title('Equalized Image')
plt.show()

# Read Color Image
img = cv2.imread('parrot.webp', cv2.IMREAD_COLOR)

# Convert BGR to HSV
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Apply Histogram Equalization to V Channel
img_hsv[:, :, 2] = cv2.equalizeHist(img_hsv[:, :, 2])

# Convert HSV to BGR
img_eq = cv2.cvtColor(img_hsv, cv2.COLOR_HSV2BGR)

# Display Equalized Color Image
plt.imshow(img_eq[:, :, ::-1])
plt.title('Equalized Color Image')
plt.show()

# Histogram of Equalized Color Image
plt.hist(img_eq.ravel(), 256, range=[0, 256])
plt.title('Histogram Equalized')
plt.show()

# Compare Original and Equalized Color Images
plt.figure(figsize=(20, 10))

plt.subplot(221)
plt.imshow(img[:, :, ::-1])
plt.title('Original Color Image')

plt.subplot(222)
plt.imshow(img_eq[:, :, ::-1])
plt.title('Equalized Color Image')

plt.show()

# Compare Original and Equalized Histograms
plt.figure(figsize=(15, 4))

plt.subplot(121)
plt.hist(img.ravel(), 256, range=[0, 256])
plt.title('Original Image Histogram')

plt.subplot(122)
plt.hist(img_eq.ravel(), 256, range=[0, 256])
plt.title('Equalized Image Histogram')

plt.show()
```
